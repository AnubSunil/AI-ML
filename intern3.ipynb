```python
df['Age_mean_imputed']=df['Age'].fillna(df['Age'].mean())
df['Age_median_imputed']=df['Age'].fillna(df['Age'].median())
```

```python
df['Monthly_Salary_mean_imputed']=df['Monthly_Salary'].fillna(df['Monthly_Salary'].mean())
df['Monthly_Salary_median_imputed']=df['Monthly_Salary'].fillna(df['Monthly_Salary'].median())
```

```python
df['Department_mode_imputed']=df['Department'].fillna(df['Department'].mode()[0])
```

```python
df['Education_mode_imputed']=df['Education'].fillna(df['Education'].mode()[0])
```

```python
df['City_mode_imputed']=df['City'].fillna(df['City'].mode()[0])
```

```python
sns.set(style="whitegrid")
plt.figure(figsize=(10,6))
department_counts = df['Department'].value_counts()
colors = sns.color_palette('pastel')[0:len(department_counts)]
plt.pie(department_counts, labels=department_counts.index, autopct='%1.1f%%', startangle=90, colors=colors)
plt.title('Distribution of Departments')
plt.ylabel('')
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoded_data=df.copy()
```

```python
label_encoder=LabelEncoder()
encoded_data['Gender_label']=label_encoder.fit_transform(encoded_data['Gender'])
```

```python
one_hot_encoded=pd.get_dummies(encoded_data['Department_mode_imputed'],prefix='Department_mode_imputed')
encoded_data=pd.concat([encoded_data,one_hot_encoded],axis=1)
```

```python
Rating_mapping={1: '1STAR',2:'2STAR',3:'3STAR',4:'4STAR',5:'5STAR'}
encoded_data['Rating_Mapped']=encoded_data['Performance_Rating'].map(Rating_mapping)
```

```python
print(encoded_data[['Gender', 'Gender_label', 'Department', 'Department_mode_imputed_Sales', 'Department_mode_imputed_IT', 'Department_mode_imputed_HR', 'Department_mode_imputed_Marketing', 'Performance_Rating', 'Rating_Mapped']].head())
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
numerical_col=['Age','Experience_Years','Monthly_Salary']
df.dropna(subset=numerical_col,inplace=True)
minmax_scaler=MinMaxScaler()
standard_scaler=StandardScaler()
df['Age_MinMax']=minmax_scaler.fit_transform(df[['Age']])
df['Age_Standard']=standard_scaler.fit_transform(df[['Age']])
```

```python
df['Experience_Years_MinMax']=minmax_scaler.fit_transform(df[['Experience_Years']])
df['Experience_Years_Standard']=standard_scaler.fit_transform(df[['Experience_Years']])
```

```python
df['Monthly_Salary_MinMax']=minmax_scaler.fit_transform(df[['Monthly_Salary']])
df['Monthly_Salary_Standard']=standard_scaler.fit_transform(df[['Monthly_Salary']])
```
